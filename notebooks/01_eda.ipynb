# Phishing Detection EDA

Minimal exploratory analysis for the processed phishing dataset.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")
```

```python
# 1) Load dataset
data_path = "../data/processed/dataset.csv"
df = pd.read_csv(data_path)
df.head()
```

```python
# 2) Shape + first rows
print(f"Dataset shape: {df.shape}")
df.head()
```

```python
# 3) Class distribution countplot
target_candidates = ["class", "target", "label"]
target_col = next((c for c in target_candidates if c in df.columns), df.columns[-1])

plt.figure(figsize=(6, 4))
sns.countplot(data=df, x=target_col)
plt.title(f"Class Distribution ({target_col})")
plt.xlabel(target_col)
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

```python
# 4) Summary statistics
df.describe()
```

```python
# 5) Correlation heatmap of numeric features
numeric_df = df.select_dtypes(include="number")
corr = numeric_df.corr()

plt.figure(figsize=(12, 8))
sns.heatmap(corr, cmap="coolwarm", center=0)
plt.title("Correlation Heatmap (Numeric Features)")
plt.tight_layout()
plt.show()
```

## Written Insight

The class distribution appears reasonably balanced, with no extreme dominance by a single class, which is favorable for supervised model training. Feature summary statistics suggest that several numeric predictors vary across broad ranges, indicating heterogeneous behavior and potential discriminative value. At the same time, some features show lower variance and may contribute less signal on their own. The heatmap indicates pockets of moderate-to-strong correlations among certain feature groups, which may be useful for interpretation but should be monitored to avoid redundancy in modeling.
