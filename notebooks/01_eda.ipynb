# Phishing Detection EDA

Minimal exploratory analysis for the processed phishing dataset.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")
```

```python
# 1) Load dataset
data_path = "../data/processed/dataset.csv"
df = pd.read_csv(data_path)
df.head()
```

```python
# 2) Shape + first rows
print(f"Dataset shape: {df.shape}")
df.head()
```

```python
# 3) Class distribution countplot
target_candidates = ["class", "target", "label", "status"]
target_col = next((c for c in target_candidates if c in df.columns), df.columns[-1])

class_counts = df[target_col].value_counts(dropna=False)
class_pct = (df[target_col].value_counts(normalize=True, dropna=False) * 100).round(2)
print("Class counts:")
print(class_counts)
print("\nClass percentages (%):")
print(class_pct)

plt.figure(figsize=(6, 4))
sns.countplot(data=df, x=target_col)
plt.title(f"Class Distribution ({target_col})")
plt.xlabel(target_col)
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

```python
# 4) Summary statistics
df.describe()
```

```python
# 5) Correlation heatmap of numeric features
numeric_df = df.select_dtypes(include="number")
corr = numeric_df.corr()

plt.figure(figsize=(12, 8))
sns.heatmap(corr, cmap="coolwarm", center=0)
plt.title("Correlation Heatmap (Numeric Features)")
plt.tight_layout()
plt.show()
```

## Written Insight

The dataset is perfectly balanced between the two target classes, with 5,715 `legitimate` samples and 5,715 `phishing` samples (50/50). This is useful for baseline modeling because metrics are less likely to be inflated by class imbalance. Summary statistics show wide ranges across several numeric URL/page features, suggesting meaningful variability and predictive signal. The correlation heatmap also indicates some correlated feature groups, which is informative for interpretation and future feature selection.
